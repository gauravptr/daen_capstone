```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Load points
points = np.loadtxt(r"C:\Users\grsha\Desktop\DAEN 460\output\underwater_scene_scan.xyz")

print(f"Loaded {len(points)} points")

# Plot
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Subsample for faster plotting if needed
if len(points) > 50000:
    idx = np.random.choice(len(points), 50000, replace=False)
    points = points[idx]

ax.scatter(points[:, 0], points[:, 1], points[:, 2], s=0.1, c=points[:, 2], cmap='viridis')

ax.set_xlabel('X (m)')
ax.set_ylabel('Y (m)')
ax.set_zlabel('Z (m)')
ax.set_title('Combined Point Cloud Scan')

plt.tight_layout()
plt.show()
```
